# Code to estimate a naive Bayes classifier

This code example converts a "rectangular" dataset into the numeric conditional probability tables for a Bayes network implementation of a multinomial naive Bayes classifier in python without use of python machine learning libraries. 

Input:  A CSV file with column labels, and enumerated values (a fixed set of strings or integers) as the data. One column is designated as the target variable, to be predicted by the other feature variables. 

The Bayes network structure consists of discrete  distributions with the target variable as the parent in the network that conditions each 

Output: A list conditional probability tables (CPTs) consisting of column markov matrices, one for each feature variable, with rows for each state of the feature variable and columns for each state of the target variable. The rows and columns are labeled by the names of the variable states. 

The target variable consists of one column since it is not conditioned by other variables. 

Algorithm

- Initialize each CPT to uniform counts using DEFAULT_COUNT. 
- Compute pivot tables pairwise with each feature and the target variable, to create CPTs.
- Compute the marginal counts for the target variable
- If NORMALIZE is true, convert each matrix so columns sum to one. 

Print a visual report of each CPT, along with it's entropy and its mutual information with the target. 

Write out a python file of the CPTs that can be imported into a python script.
