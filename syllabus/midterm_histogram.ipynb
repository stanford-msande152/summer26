```python
fig, axs = plt.subplots(1, 2, figsize=(10, 4))
plt.suptitle('Distribution of Grades by School Type (HS vs. Non-HS)')
sns.histplot(
    data= grades[grades['school'] != 'HS'],
    x="raw",
    ax=axs[0],
    kde=True,
    bins=18)
sns.histplot(
    data= grades[grades['school'] == 'HS'],
    x="raw",
    ax=axs[1],
    kde=True,
    bins=18,
    color='DarkOrange')
# fig.title('Distribution of Grades by School Type (HS vs. Non-HS)')
axs[0].set_title('Non-HS Grades')
axs[1].set_title('HS Grades')
axs[0].set_xlabel('Grade')
axs[1].set_xlabel('Grade')
axs[0].set_ylabel('Frequency')
fig.show()
```
