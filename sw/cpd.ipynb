```python
import numpy as np
import numpy.typing as npt
from bokeh.plotting import figure, show, save
from bokeh.layouts import column
from bokeh.models import ColumnDataSource, FixedTicker, MultiLine, Grid, Plot, Label, LinearAxis, Legend, Span, VBar, Step
from bokeh.io import output_notebook
from bokeh import palettes
output_notebook()
```

```python
# Example of a uniform outcome distribution. 

pts=8
rng = np.random.default_rng()
uniform_p = rng.random(pts)
uniform_p = uniform_p/uniform_p.sum()
uniform_u = np.linspace(30, 100, pts)
uniform_p2 = uniform_p2/uniform_p2.sum()
uniform_u2 = np.linspace(10, 80, pts)
```

```python
# A function to display a step plot for a cumulative probability distribution using bokeh

def cpd(p: npt.ArrayLike, u: npt.ArrayLike):
    p = np.asarray(p, dtype=float)
    u = np.asarray(u, dtype=float)
    # Sort by utility so unsorted inputs still produce a monotonic CPD
    order = np.argsort(u)
    u = u[order]
    p = p[order]
    cdf = np.cumsum(p)

    cdf_v = np.concatenate(([0,0], cdf))
    u_v = np.concatenate(([0, u[0]], u))

    # Compute expected utility
    eu = np.multiply(p, u).sum()
    print(f'Expected Utility = {eu:.2}')

    fig = figure(
        title="Prospect Cumulative Probability Distribution",
        width=400, height=300,
        background_fill_color="#fafafa",
        x_range=(0, 100), y_range=(0, 1),
        x_axis_label='Utility',
        y_axis_label='Probability',
        toolbar_location=None  # eliminate the default tool panel on the right
    )

    # Build the CPD as a step function. The Step glyph takes parallel lists x and y
    # and renders a step plot: the value is constant between consecutive x values
    # and changes at the next x. With mode='after' the y value at x_i is held until
    # x_{i+1}, which is the standard convention for an empirical CDF.
    cpd_src = ColumnDataSource(data=dict(x=u_v, y=cdf_v))
    # print(cdf_v)
    # print(u_v)
    step_glyph = Step(x="x", y="y", line_width=2, line_color="navy", mode="after")
    fig.add_glyph(cpd_src, step_glyph)

    # Scatter of the cusp points (the (u, CDF) values where the step jumps)
    cusp_src = ColumnDataSource(data=dict(x=u, y=cdf))
    fig.scatter(x="x", y="y", source=cusp_src, size=6, color="navy")

    vline = Span(location=eu, dimension='height', line_color='navy', line_width=3, line_alpha= 0.4)
    # fig.add_glyph(cusp_src, vline)
    fig.add_layout(vline)
    my_label = Label(x=eu, y=0.06, text=f'EU = {eu:.1f}', text_font_size="9pt", text_color="navy")
    fig.add_layout(my_label)

    return fig
```

```python
def cpd2(p: npt.ArrayLike, u: npt.ArrayLike, p2: npt.ArrayLike, u2: npt.ArrayLike):

    eu,cdf_v,u_v = to_cpd(p, u)
    eu2,cdf_v2,u_v2 = to_cpd(p2, u2)

    fig = figure(
        title="Prospect Cumulative Probability Distribution",
        width=400, height=300,
        background_fill_color="#fafafa",
        x_range=(0, 100), y_range=(0, 1),
        x_axis_label='Utility',
        y_axis_label='Probability',
        toolbar_location=None  # eliminate the default tool panel on the right
    )

    fig = plot_cdf(eu, cdf_v, u_v, fig, col='red')

    fig = plot_cdf(eu2, cdf_v2, u_v2, fig)

    return fig

def plot_cdf(eu, cdf_v, u_v, fig, col='navy'):
    # Build the CPD as a step function. The Step glyph takes parallel lists x and y
    # and renders a step plot: the value is constant between consecutive x values
    # and changes at the next x. With mode='after' the y value at x_i is held until
    # x_{i+1}, which is the standard convention for an empirical CDF.
    cpd_src = ColumnDataSource(data=dict(x=u_v, y=cdf_v))
    # print(cdf_v)
    # print(u_v)
    step_glyph = Step(x="x", y="y", line_width=2, line_color=col, mode="after")
    fig.add_glyph(cpd_src, step_glyph)

    # Scatter of the cusp points (the (u, CDF) values where the step jumps)
    cusp_src = ColumnDataSource(data=dict(x=u_v, y=cdf_v))
    fig.scatter(x="x", y="y", source=cusp_src, size=6, color=col)

    vline = Span(location=eu, dimension='height', line_color=col, line_width=3, line_alpha= 0.4)
    # fig.add_glyph(cusp_src, vline)
    fig.add_layout(vline)
    my_label = Label(x=eu, y=0.06, text=f'EU = {eu:.1f}', text_font_size="9pt", text_color=col)
    fig.add_layout(my_label)
    return fig

def to_cpd(p, u):
    p = np.asarray(p, dtype=float)
    u = np.asarray(u, dtype=float)
    # Sort by utility so unsorted inputs still produce a monotonic CPD
    order = np.argsort(u)
    u = u[order]
    p = p[order]
    cdf = np.cumsum(p)

    cdf_v = np.concatenate(([0,0], cdf))
    u_v = np.concatenate(([0, u[0]], u))

    # Compute expected utility
    eu = np.multiply(p, u).sum()
    print(f'Expected Utility = {eu:.2f}')
    return eu,cdf_v,u_v
```

```python
# stochastic dominance
show(cpd2(uniform_p, uniform_u, uniform_p2,uniform_u2))
```
