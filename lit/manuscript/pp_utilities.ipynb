```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
values = [50, 40, 20, 90, 0 ,100]
pref_probs = [0.67, 0.57, 0.32, 0.95, 0.0, 1.0]
labels = ['O-R', 'P-R',  'I-S', 'I-R',  'P-S', 'O-S']
energy_alts = [ 'Wind generation, base' ,
 'Solar plant, base'     ,
 'Battery storage, peak' ,
 'Battery storage, base' ,
 'Solar plan, peak'      ,
 'Wind generation, peak' ]
energy_alts = list(reversed(energy_alts))
indexes_by_magnitude = sorted(range(len(values)), key=values.__getitem__)
values = [values[v] for v in indexes_by_magnitude]
pref_probs = [pref_probs[v] for v in indexes_by_magnitude]
indexes_by_magnitude
```

```python
### Kim's utility function

def kim_utility(x):
    return (4/3)*(1 - pow(0.5, x/50))
```

```python
v = np.linspace(0, 100)
ku = [kim_utility(z) for z in v]
```

```python

plt.figure(figsize=(8, 4))
plt.plot([values[k] for k in indexes_by_magnitude], 
         [pref_probs[k] for k in indexes_by_magnitude], marker='o', linestyle=':')
plt.plot(v, ku, linestyle='-', color = 'grey')
plt.ylim(-0.1, 1.2)
for x, y, label in zip(values, pref_probs, labels):
    plt.annotate(label, (x, y), textcoords="offset points", xytext=(0, 8), ha='center')

plt.xlabel('Values')
plt.ylabel('Preference Probabilities')
plt.title('Risk averse utility function')
plt.grid(True, alpha=0.3)
plt.show()
```

```python

# Negative value, shows insurance 
neg_v = np.linspace(-80, 50)
neg_ku = [kim_utility(z) for z in neg_v]
plt.figure(figsize=(8, 4))
plt.plot(neg_v, neg_ku, linestyle='-', color = 'black')
plt.plot([-80,0], [kim_utility(-80),0], linestyle = ':')
plt.ylim(-2.9, 0.8)
plt.axhline(y=0, color='tan', linewidth = 0.5)
plt.axvline(x=0, color='tan', linewidth = 0.5)
plt.xlabel('$V')
plt.ylabel('u(V)')
plt.title('Risk aversion explains insurance premiums')
plt.grid(True, alpha=0.3)
plt.show()
```
