```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
values = [50, 40, 20, 90, 0 ,100]
pref_probs = [0.67, 0.57, 0.32, 0.95, 0.0, 1.0]
labels = ['P-R', 'P-S', 'I-R', 'I-S', 'O-R', 'O-S']

indexes_by_magnitude = sorted(range(len(values)), key=values.__getitem__)
indexes_by_magnitude
```

```python
### Kim's utility function

def kim_utility(x):
    return (4/3)*(1 - pow(0.5, x/50))
```

```python
v = np.linspace(0, 100)
ku = [kim_utility(z) for z in v]
```

```python

plt.figure(figsize=(8, 4))
plt.plot([values[k] for k in indexes_by_magnitude], 
         [pref_probs[k] for k in indexes_by_magnitude], marker='o', linestyle=':')
plt.plot(v, ku, linestyle='-', color = 'grey')
plt.ylim(-0.1, 1.2)
for x, y, label in zip(values, pref_probs, labels):
    plt.annotate(label, (x, y), textcoords="offset points", xytext=(0, 8), ha='center')

plt.xlabel('Values')
plt.ylabel('Preference Probabilities')
plt.title('Risk averse utility function')
plt.grid(True, alpha=0.3)
plt.show()
```
